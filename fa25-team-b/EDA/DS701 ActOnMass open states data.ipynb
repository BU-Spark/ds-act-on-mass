```python
headers = {"X-API-KEY": API_KEY}
output_file = "massachusetts_bills.csv"

max_pages = 50  

with open(output_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Bill ID", "Chamber", "Title", "First Action", "Latest Action", "Latest Description", "Session", "Bill URL"])

    total_rows = 0

    for page in range(1, max_pages + 1):
        params = {
            "jurisdiction": "Massachusetts",
            "per_page": 20,  #API maximum limit 
            "page": page
        }

        response = requests.get(BASE_URL, params=params, headers=headers)
        if response.status_code != 200:
            break

        data = response.json()
        results = data.get("results", [])
        if not results:
            break

        for bill in results:
            writer.writerow([
                bill.get("identifier"),
                bill.get("from_organization", {}).get("name"),
                bill.get("title"),
                bill.get("first_action_date"),
                bill.get("latest_action_date"),
                bill.get("latest_action_description"),
                bill.get("session"),
                bill.get("openstates_url")
            ])
            total_rows += 1

        time.sleep(0.5) 
```
