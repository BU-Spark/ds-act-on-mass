This method works for the one document I'm using to test, but unfortunately the PDFs aren't all reading in consistently, so this may not be useful for generalizing.

I combined the above code into a single function, but again, it pretty much only works for the one document. Will need to investigate further to figure out how to get the expected results more consistently.

```python
def parse_pdf(file_path):
    doc = pymupdf.open(file_path)
    page1 = doc[0]
    tables = page1.find_tables(strategy = "text")
    table1 = tables[0]
    df = table1.to_pandas()
    df_new = df.rename(columns = {df.columns[0]: 0, df.columns[1]: 1, df.columns[2]: 2, df.columns[3]: 3, df.columns[4]: 4, df.columns[5]: 5, df.columns[6]: 6, df.columns[7]: 7})

    votes = []
    names = []
    for column in df_new:
        if (column + 2) % 2 == 0:
            for row in df_new[column][3:43]:
                if row != "": #Dropping empty rows/values
                    votes.append(row)
        else:
            for row in df_new[column][3:43]:
                if row != "": #Dropping empty rows/values
                    names.append(row)

    data = list(zip(names, votes))
    return (pd.DataFrame(data, columns = ["Name", "Vote"]))
```
