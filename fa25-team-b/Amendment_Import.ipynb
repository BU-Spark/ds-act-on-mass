```python

import numpy as np
import pandas as pd
import requests
```

Import Data

```python
df = pd.read_csv("data/data_Bills_194.csv")
```

```python
print(df["Amendments"].describe())
df2 = df.copy()
```

Get all unique amendments and make sure to only get amendments

```python



unique_df = df2["Amendments"].drop_duplicates()

unique_df.replace("[]", np.nan, inplace=True) #Replace empty with NaN
unique_df.dropna(inplace=True) #Remove NaN
unique_df= unique_df[~unique_df.str.startswith("4")]
unique_df= unique_df[~unique_df.str.startswith("5")] # Remove 400 and 500 errors that might have shown up.
unique_indices =unique_df.index
```

Display shape of current data

```python
df3 = pd.DataFrame(eval(df2["Amendments"][139]))
print(df3)
print(df2["Amendments"][139])
```

Check for duplicate Amendments and design the dataframe to be a table for Looker

```python
skipped = 0
#There are duplicate amendments written twice. need to fix that somehow.

df4 = pd.DataFrame(eval((df2["Amendments"][unique_indices[0]])))
df3 = pd.DataFrame(columns=df4.columns)
for i in unique_indices:
    
    try:
        df4 = pd.DataFrame(eval((df2["Amendments"][i])))
        df3=pd.concat([df3,df4], ignore_index = True)
    
        
    except:
        skipped +=1
print(f"Skipped {skipped} numbers (any missing are errors)")
print(len(df3))

#Remove duplicate amendments
df3.drop_duplicates(subset=['AmendmentNumber', 'ParentBillNumber'], inplace=True) 

df3.drop(0, axis = 1, inplace=True)
print(df3.head())
print(len(df3))

```

Save File

```python
df3.to_csv("data/amendments.csv", index = False)
```
