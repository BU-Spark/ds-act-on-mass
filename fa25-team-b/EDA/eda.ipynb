# House & Senate Journals + Roll Calls EDA
## Massachusetts Legislature - Basic Analysis

## Setup

```python
import requests
import pandas as pd
import json

BASE_URL = "https://malegislature.gov/api"
GENERAL_COURT = 193  # Change to 194 if needed

print(f"Analyzing General Court {GENERAL_COURT}")
```

## Analysis 1: Roll call

```python
# Get roll calls for Senate document S2538
doc_id = "S2538"
rollcalls = requests.get(
    f"{BASE_URL}/Documents/{doc_id}/RollCalls"
).json()

print("\nROLL CALLS FOR DOCUMENT", doc_id)
print("=" * 60)

if isinstance(rollcalls, list):
    print(f"Total Roll Calls: {len(rollcalls)}")
    
    if rollcalls:
        print(f"\nAvailable Fields: {list(rollcalls[0].keys())}")
        print(f"\nAll Roll Calls:")
        for rc in rollcalls:
            print(f"  - {rc}")
else:
    print("No roll calls or unexpected format")
    print(rollcalls)
```

## Get Detailed Roll Call Information

```python
# Get detailed roll call 55 for Senate
roll_call_num = 55
detailed_rollcall = requests.get(
    f"{BASE_URL}/GeneralCourts/194/Branches/Senate/RollCalls/{roll_call_num}"
).json()

print(f"\n\nDETAILED ROLL CALL #{roll_call_num}")
print("=" * 60)
print(f"Date: {detailed_rollcall.get('Date')}")
print(f"Description: {detailed_rollcall.get('Description')}")

# Check available fields
print(f"\nAvailable Fields: {list(detailed_rollcall.keys())}")

# Check for Yeas/Nays structure
if 'Yeas' in detailed_rollcall:
    yeas = detailed_rollcall['Yeas']
    nays = detailed_rollcall['Nays']
    
    print(f"\nVote Totals:")
    print(f"  Yeas: {len(yeas)}")
    print(f"  Nays: {len(nays)}")
    
    if yeas:
        print(f"\nFirst Yea Vote:")
        print(f"  Name: {yeas[0].get('Name')}")
        print(f"  Party: {yeas[0].get('Party')}")
        print(f"  District: {yeas[0].get('District')}")
```

## Test Multiple Senate Documents

```python
# Test several Senate documents
test_docs = ["S2538", "S2539", "S2540", "S2541", "S2542"]

doc_rollcall_summary = []

for doc in test_docs:
    try:
        rc = requests.get(f"{BASE_URL}/Documents/{doc}/RollCalls").json()
        count = len(rc) if isinstance(rc, list) else 0
        doc_rollcall_summary.append({
            'Document': doc,
            'Roll Call Count': count,
            'Has Roll Calls': count > 0
        })
        print(f"{doc}: {count} roll calls")
    except Exception as e:
        print(f"{doc}: Error - {e}")

print("\nDocuments with Roll Calls:")
for item in doc_rollcall_summary:
    if item['Has Roll Calls']:
        print(f"  ✓ {item['Document']}: {item['Roll Call Count']} roll calls")
```

## Summary

```python
total_senate_with_rc = sum(1 for item in doc_rollcall_summary if item['Has Roll Calls'])

print(f"Senate Documents Tested: {len(test_docs)}")
print(f"Senate Documents with Roll Calls: {total_senate_with_rc}")

print("\nKey Insights:")
print("- Use /Documents/{docNumber}/RollCalls to get roll calls for a specific bill")
print("- Not all documents have roll calls")
print("- Roll call endpoint returns a list of roll call references")
```
