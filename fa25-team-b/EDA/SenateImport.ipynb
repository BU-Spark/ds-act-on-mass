Try to get a senate roll call report.

```python
import requests #Using requests as a package for getting the https files given by the api
```

```python
document1 = requests.get("https://malegislature.gov/api/Documents/S2538")
#This is asking for one specific document. Change S2538 to a specific document to change the document
# It is possible that this only reads documents from the most recent general court (194 as of now)
# If this only read that then change link after /api to /GeneralCourts/194/Documents/S2538 to get the same document
# in this case change 194 to your general court and S2538 to your document for other options.
```

```python
print(document1.json())
# This displays it as one json file
```

```python
rollcalls = requests.get("https://malegislature.gov/api/Documents/S2538/RollCalls")
# This just requests the RollCalls within one specific document they can also just be read from that document
```

```python
# Display as one json file
print(rollcalls.json())
```

```python
# go to one of the rollcalls in prior
rollcall1 = requests.get("https://malegislature.gov/api/GeneralCourts/194/Branches/Senate/RollCalls/55")
```

```python
# We can now read the individual yeas and nays for that amendment to the previous document
# Note this does not reference the earlier document in any way and you can't easily trace it backwards
print(rollcall1.json())
```

```python
# Let's instead inside of the document look at one of the amendments.
amendment6 = requests.get("https://malegislature.gov/api/GeneralCourts/194/Documents/S2538/Branches/Senate/Amendments/6")
```

```python
print(amendment6.json())
# This amendment links to the roll call info and contains a description. 
# This can be used to find what a roll call was used for and to explain what was voted for.

```

```python

print(type(amendment6.json()))
print(type(amendment6.text)) 
# It should be easier to work with a dictionary than a string so preferably use .json()
#Note if there is any error in the request it will not be in a json readable format and give an error, 
#it will state an error in text format though
```

```python
# Let's request info about the person who sponsored that amendment 6
person1 = requests.get("https://malegislature.gov/api/GeneralCourts/194/LegislativeMembers/MCM0")
```

```python
# This contains information about the individual and other bills they've sponsored or cosponsored
# This does not contain any info on any votes of theirs.
print(person1.json())
```
