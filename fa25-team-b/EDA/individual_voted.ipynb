```python
def get_bills_by_senator(csv_path, senator_id):
    """
    Returns all bills sponsored or voted on by a given senator.
    """
    # Load the dataset
    df = pd.read_csv(csv_path)
    
    # Bills the senator sponsored
    sponsored_bills = df.loc[df['PrimarySponsor.Id'] == senator_id, 'BillNumber'].dropna().tolist()

    # 2️⃣ Bills the senator voted on (if 'RollCalls' field contains JSON/text with senator IDs)
    voted_bills = []
    if 'RollCalls' in df.columns:
        # Check which rows contain this senator_id in the RollCalls column (string search)
        mask = df['RollCalls'].astype(str).str.contains(senator_id, na=False)
        voted_bills = df.loc[mask, 'BillNumber'].dropna().tolist()

    # Combine and deduplicate
    all_bills = sorted(set(sponsored_bills + voted_bills))
    return all_bills
```

```python
bills = get_bills_by_senator("data/data_Bills_194.csv", "JCD1")
print(bills)
# Output: List of unique BillNumbers associated with senator ID "JCD1"
```
